## **Problem Statement**

The banking industry continually seeks to optimize marketing strategies, particularly for products such as term deposits. Traditional direct marketing campaigns often conducted via phone calls, face significant challenges.

#### **1. Low Conversion Rates**
Many marketing efforts result in low subscription rates for term deposits, leading to inefficient resource allocation and missed revenue opportunities. The manual, often untargeted, approach leads to high volume of contacts for small number of successful subscriptions.

#### **2. Wasted Resources**
Significant resources (time, personnel & operational costs) are expended on contacting individuals who have little to no interest in subscribing, which reduces overall efficiency & profitability of marketing campaigns.

#### **3. Lack of Targeted Approach**
Without predictive understanding of customer behavior, banks struggle to identify & prioritize potential clients most likely to convert, which often leads to a 'spray & pray' approach rather than data-driven, targeted strategy.

### **Objective**
The primary objective is to develop robust predictive model which can accurately identify prospective customers, most likely to subscribe to term deposit. By leveraging historical marketing & demographic data, this model will enable the bank to.
*   **Improve Marketing Efficiency**: Focus efforts on high-potential leads, reducing marketing costs & increasing return on investment.
*   **Enhance Customer Acquisition**: Boost term deposit subscription rates through more personalized & effective campaigns.
*   **Gain Actionable Insights**: Uncover key factors influencing client's decision to subscribe, allowing for development of more tailored product offerings & refined marketing strategies.

## **Business Goal**
The primary business goal is to develop predictive model which can accurately identify potential customers, most likely to subscribe for term deposit, which will enable the bank to
*   **Optimize Marketing Campaigns**: Focus resources on high-potential leads, reducing marketing costs & increasing conversion rates.
*   **Improve Customer Acquisition**: Enhance effectiveness of term deposit campaigns, leading to higher subscription numbers.
*   **Understand Customer Behavior**: Gain insights into factors influencing client's decision to subscribe term deposit, allowing for more tailored product offerings & marketing strategies.

## **Solution Approach**

To achieve stated business goals & to address problem of low term deposit subscription rates, structured data science approach will be followed. This approach involves several key phases aimed at building and deploying predictive model.

1.  **Data Preprocessing**: The initial phase focuses on preparing raw data for analysis & modeling, which includes cleaning, handling missing values, converting data types & ensuring data quality.

2.  **Exploratory Data Analysis (EDA)**: Through various statistical & visualization techniques, this phase aims to understand dataset's characteristics, uncover patterns, identify relationships between variables & detect potential anomalies inlcuding understanding the distribution of target variable & identifying any class imbalance.

3.  **Feature Engineering**: This phase involves creating new features or transforming existing ones to enhance predictive power of models, which also includes encoding categorical variables & scaling numerical features as required by machine learning algorithms.

4.  **Modeling**: In this phase, various machine learning models including baseline model (Logistic Regression) & several advanced models (e.g., Random Forest, Gradient Boosting), will be developed & trained. The goal is to identify models capable of accurately predicting client subscription behavior.

5.  **Tuning**: This phase involves optimizing hyperparameters of selected models using systematic search strategies & cross-validation to maximize performance on unseen data, with focus on metrics relevant to imbalanced datasets.

6.  **Evaluation**: The final phase involves rigorously evaluating best-performing model on held-out test set using appropriate metrics (e.g., AUC-ROC, F1-Score, Precision, Recall) to ensure generalization capability & interpretability for actionable business insights.

### 1.14 Expected Challenges

*   **Class Imbalance**: Target variable `y` (subscription to term deposit) is almost certainly highly imbalanced. Typically, only small fraction of customers subscribe to such products which poses significant challenge as models trained on imbalanced data tend to be biased towards majority class, leading to poor performance on minority (positive) class. To address it, techniques such as **SMOTE (Synthetic Minority Over-sampling Technique)**, **class weighting**, or **under-sampling majority class** will be considered during model training. Evaluation will heavily rely on metrics robust to imbalance such as **F1-score, Precision, Recall, and AUC-ROC**, rather than simple accuracy.

*   **Categorical Variable Handling**: `df_bank_additional_full` contains 11 categorical features (`job`, `marital`, `education`, `default`, `housing`, `loan`, `contact`, `month`, `day_of_week`, `poutcome`, `y`). Many of these (e.g., `job`, `education`, `default`, `housing`, `loan`, `poutcome`) contain an 'unknown' category, to be carefully handled. Strategies include.
    *   **One-Hot Encoding**: Suitable for nominal categorical variables, but can lead to high dimensionality if categories are numerous.
    *   **Label Encoding**: Appropriate for ordinal variables, but can impose an artificial order if not inherently present.
    *   **Target Encoding**: Can capture target-related information, but susceptible to overfitting.
    *   **Handling 'unknown'**: This can be treated as a separate valid category or imputed based on other features.

*   **Feature `duration` (Data Leakage)**: The `duration` feature, representing last contact duration in seconds, is highly problematic. The dataset description states that `duration = 0` implies `y = 'no'`, which creates direct, unrealistic correlation, as outcome of call is only known *after* call. Including such feature directly in predictive model would lead to **data leakage**, resulting in overly optimistic & untrustworthy model. To mitigate.
    *   The `duration` feature might be **removed** entirely.
    *   Alternatively, transformed (e.g., into a binary indicator `contact_made` or `contact_successful`), or used in combination with other features, but with extreme caution to prevent leakage. The primary goal is to predict subscription *before* call, not *during* or *after*.

*   **Feature `pdays` (Handling -1 values)**: The `pdays` feature indicates 'days since last contact from a previous campaign', with -1 meaning customer was not previously contacted. #### 2.2.3.1 Z-score method for extreme values

The Z-score method measures how many standard deviations an element is from the mean. A common threshold for identifying extreme outliers is a Z-score greater than 3 or less than -3. This method is effective for normally distributed data but can be less reliable for skewed distributions.

```python
# Z-score method for extreme values
from scipy import stats

# Define the numerical columns for outlier detection (excluding pdays and duration)
numeric_cols = ['age', 'campaign', 'previous', 'emp.var.rate', 'cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']

print("=== Z-Score Method (|Z| > 3) ===")
for col in numeric_cols:
    z_scores = np.abs(stats.zscore(df_bank_additional_full[col]))
    extreme_outliers = df_bank_additional_full[z_scores > 3]
    print(f"\n{col}:")
    print(f"  Extreme outliers: {len(extreme_outliers)} ({len(extreme_outliers)/len(df_bank_additional_full)*100:.2f}%) ")
    if len(extreme_outliers) > 0:
        print(f"  Max value: {df_bank_additional_full[col].max():.2f}")
        print(f"  Min value: {df_bank_additional_full[col].min():.2f}")

# Visual representation of Z-score method
plt.figure(figsize=(18, 10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(2, 4, i)

    # Calculate Z-scores
    z_scores = stats.zscore(df_bank_additional_full[col])

    # Create histogram of Z-scores
    plt.hist(z_scores, bins=30, alpha=0.7, color='lightcoral', edgecolor='black')

    # Add threshold lines
    plt.axvline(3, color='red', linestyle='--', linewidth=2, label='Z = +3 (Upper Threshold)')
    plt.axvline(-3, color='red', linestyle='--', linewidth=2, label='Z = -3 (Lower Threshold)')
    plt.axvline(0, color='blue', linestyle='-', linewidth=2, label='Mean (Z = 0)')

    # Highlight extreme outliers
    extreme_outliers_data = z_scores[np.abs(z_scores) > 3]
    if len(extreme_outliers_data) > 0:
        plt.scatter(extreme_outliers_data, [0]*len(extreme_outliers_data), color='red', s=50, alpha=0.8, label=f'Extreme Outliers: {len(extreme_outliers_data)}')

    plt.title(f'Z-Score Method: {col}')
    plt.xlabel('Z-Score')
    plt.ylabel('Frequency')
    plt.legend(fontsize=8)
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Additional visualization: Original values with Z-score thresholds
plt.figure(figsize=(18, 10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(2, 4, i)

    # Calculate Z-score thresholds in original scale
    mean_val = df_bank_additional_full[col].mean()
    std_val = df_bank_additional_full[col].std()
    upper_threshold = mean_val + 3 * std_val
    lower_threshold = mean_val - 3 * std_val

    # Create histogram of original values
    plt.hist(df_bank_additional_full[col], bins=30, alpha=0.7, color='lightgreen', edgecolor='black')

    # Add threshold lines
    plt.axvline(upper_threshold, color='red', linestyle='--', linewidth=2, label=f'Upper: {upper_threshold:.2f}')
    plt.axvline(lower_threshold, color='red', linestyle='--', linewidth=2, label=f'Lower: {lower_threshold:.2f}')
    plt.axvline(mean_val, color='blue', linestyle='-', linewidth=2, label=f'Mean: {mean_val:.2f}')

    # Highlight extreme outliers
    z_scores = np.abs(stats.zscore(df_bank_additional_full[col]))
    extreme_outliers_original = df_bank_additional_full[z_scores > 3]
    if len(extreme_outliers_original) > 0:
        plt.scatter(extreme_outliers_original[col], [0]*len(extreme_outliers_original), color='red', s=50, alpha=0.8, label=f'Extreme Outliers: {len(extreme_outliers_original)}')

    plt.title(f'Z-Score Method (Original Scale): {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.legend(fontsize=8)
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

#### 2.2.3.2 Statistical outlier detection using IQR method

The Interquartile Range (IQR) method is a robust way to detect outliers, especially when data is not normally distributed. It identifies outliers as values that fall below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR`, where Q1 is the first quartile, Q3 is the third quartile, and IQR is the difference between Q3 and Q1.

```python
# Statistical outlier detection using IQR method
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

# Define the numerical columns for outlier detection (excluding pdays and duration)
numeric_cols = ['age', 'campaign', 'previous', 'emp.var.rate', 'cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']

print("=== IQR Method Outlier Detection ===")
for col in numeric_cols:
    outliers, lower, upper = detect_outliers_iqr(df_bank_additional_full, col)
    print(f"\n{col}:")
    print(f"  Range: {lower:.2f} to {upper:.2f}")
    print(f"  Outliers: {len(outliers)} ({len(outliers)/len(df_bank_additional_full)*100:.2f}%)")

# Visual representation of IQR method
plt.figure(figsize=(18, 10))
for i, col in enumerate(numeric_cols, 1):
    plt.subplot(2, 4, i)

    # Calculate IQR bounds
    Q1 = df_bank_additional_full[col].quantile(0.25)
    Q3 = df_bank_additional_full[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Create histogram with IQR bounds
    plt.hist(df_bank_additional_full[col], bins=30, alpha=0.7, color='skyblue', edgecolor='black')
    plt.axvline(lower_bound, color='red', linestyle='--', linewidth=2, label=f'Lower Bound: {lower_bound:.2f}')
    plt.axvline(upper_bound, color='red', linestyle='--', linewidth=2, label=f'Upper Bound: {upper_bound:.2f}')
    plt.axvline(Q1, color='green', linestyle='-', linewidth=2, label=f'Q1: {Q1:.2f}')
    plt.axvline(Q3, color='green', linestyle='-', linewidth=2, label=f'Q3: {Q3:.2f}')

    # Highlight outliers
    outliers = df_bank_additional_full[(df_bank_additional_full[col] < lower_bound) | (df_bank_additional_full[col] > upper_bound)]
    if len(outliers) > 0:
        plt.scatter(outliers[col], [0]*len(outliers), color='red', s=50, alpha=0.8, label=f'Outliers: {len(outliers)}')

    plt.title(f'IQR Method: {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.legend(fontsize=8)
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 2.2.6 Handling `pdays` Feature

The `pdays` feature, representing 'days since last contact from previous campaign', contains special value of `999` (originally `-1` before encoding) which indicates client was not previously contacted. The value is not true numerical count of days & requires specific handling to prevent misinterpretation by predictive models, which can be addressed in 2 ways.

1.  **Create a Binary Indicator Feature**: A new feature, `was_previously_contacted`, will be created, which will be `1` if `pdays < 999` (meaning client *was* previously contacted) and `0` if `pdays = 999` (meaning the client was *not* previously contacted).
2.  **Transform `pdays` Value**: For instances where `pdays = 999`, value will be replaced with `0`, to ensure all values in `pdays` column are numerical & meaning of 'not previously contacted' is captured by new `was_previously_contacted` binary feature. This transformation maintains consistent numerical scale for `pdays` where `0` means either no previous contact (due to transformation) or contact on same day (less likely in this context for prior campaigns), while `was_previously_contacted` explicitly differentiates between clients who have history of prior campaigns and those who don't.

```python
# Create a binary indicator for 'was_previously_contacted'
# If pdays is 999 (meaning not previously contacted), set to 0, otherwise 1.
df_bank_additional_full['was_previously_contacted'] = df_bank_additional_full['pdays'].apply(lambda x: 0 if x == 999 else 1)

# Replace 999 in 'pdays' with 0
df_bank_additional_full['pdays'] = df_bank_additional_full['pdays'].replace(999, 0)

print("Handling for 'pdays' feature completed.")
print("New feature 'was_previously_contacted' created.")
print("'pdays' 999 values replaced with 0.")

print("\nFirst 5 rows with updated 'pdays' and new 'was_previously_contacted' feature:")
display(df_bank_additional_full[['pdays', 'was_previously_contacted']].head())
```

#### Summary

The original `pdays` column has had its `999` values replaced with `0`, providing consistent numerical feature where `0` now represents no previous contact or same-day contact, which ensures that model can correctly interpret absence of prior contact without misinterpreting `999` as large number of days, while leveraging predictive power of actual days since last contact for those who were previously contacted.

### 2.2.7 Interaction Features

Interaction features created by combining two or more existing features to capture non-linear relationships & potentially reveal new patterns that individual features might not show. These can enhance predictive power of model by reflecting how certain characteristics influence each other. The following interaction features can be created.

1.  **`age_education_interaction`**: Feature combines `age` & `education`, hypothesis is that combination of client's age & their education level might have distinct impact on their likelihood to subscribe to term deposit. For example, older individuals with higher education might respond differently than younger individuals with same education level, or older individuals with lower education.

2.  **`campaign_previous_interaction`**: Feature combines `campaign` (number of contacts in current campaign) and `previous` (number of contacts before this campaign). TInteraction aims to capture 'cumulative contact' or 'engagement intensity' metric. A client contacted many times in current campaign who also had many previous contacts might represent different segment than someone with few contacts overall & this combined metric could be more informative than either feature alone. This could also highlight patterns such as a client being unresponsive despite multiple contacts across campaigns.

### 2.2.9 Adding New Value-Adding Features

Based on the suggestions, five new features have been engineered to capture more nuanced patterns and relationships within the data:

1.  **Economic Health Index**: A composite score reflecting the overall economic conditions relevant to the marketing campaigns.
2.  **Recency Bins for `pdays`**: Categorizes `pdays` (days since last contact) into bins, capturing contact recency more effectively.
3.  **Number of Unknowns**: Counts the occurrences of 'unknown' values in key categorical features, indicating data availability or privacy.
4.  **Age Group Categories**: Segments `age` into predefined groups for better capturing age-related patterns.
5.  **Marketing Efficiency Ratio**: A ratio of previous contacts to current campaign contacts, indicating past marketing effectiveness.

These features aim to aggregate existing information into more concise or meaningful forms, or to capture non-linear relationships that might otherwise be missed.

```python
# Import PCA for dimensionality reduction
from sklearn.decomposition import PCA

# 1. Economic Health Index (using PCA on socio-economic indicators)
# These features are already scaled using RobustScaler, so PCA can be applied directly.
economic_cols = ['emp.var.rate', 'cons.price.idx', 'cons.conf.idx', 'euribor3m', 'nr.employed']

# Ensure all economic_cols exist before applying PCA
existing_economic_cols = [col for col in economic_cols if col in df_bank_additional_full.columns]

if existing_economic_cols:
    pca = PCA(n_components=1, random_state=42) # Using random_state for reproducibility
    df_bank_additional_full['economic_health_index'] = pca.fit_transform(df_bank_additional_full[existing_economic_cols])
    print(f"'Economic_Health_Index' created using {len(existing_economic_cols)} columns.")
else:
    print("Warning: No economic columns found to create 'Economic_Health_Index'.")
```

```python
# Re-executing the fixed 'Recency Bins for pdays' cell
# 2. Recency Bins for `pdays`
# This feature categorizes `pdays` values (where 0 means no previous contact, and >0 means days since last contact)

# After previous transformations, 'pdays' has 0 for 'no previous contact' (originally 999 or -1)
# and actual days (1-27) for previously contacted clients.
# Define bins to cover the actual range of pdays (0 to max_pdays)
min_pdays_val = df_bank_additional_full['pdays'].min() # Should be 0
max_pdays_val = df_bank_additional_full['pdays'].max() # Should be 27

# Bins:
# (-1, 0] -> 'No Previous Contact'
# (0, 7]  -> 'Very Recent (1-7 days)'
# (7, max_pdays_val] -> 'Recent (>7 days)'
bins_pdays_robust = [min_pdays_val - 1, min_pdays_val, 7, max_pdays_val + 1]
labels_pdays_robust = ['No Previous Contact', 'Very Recent (1-7 days)', 'Recent (>7 days)']

# Adjust labels if 'Very Recent' range is not applicable (e.g., if max_pdays_val is 0)
if max_pdays_val <= 0:
    bins_pdays_robust = [min_pdays_val - 1, min_pdays_val]
    labels_pdays_robust = ['No Previous Contact']
elif max_pdays_val <= 7:
    bins_pdays_robust = [min_pdays_val - 1, min_pdays_val, max_pdays_val + 1]
    labels_pdays_robust = ['No Previous Contact', f'Very Recent (1-{max_pdays_val} days)']

df_bank_additional_full['pdays_recency_bins'] = pd.cut(df_bank_additional_full['pdays'], bins=bins_pdays_robust, labels=labels_pdays_robust, right=True)

# Ordinal encode the new categorical 'pdays_recency_bins' column
encoder_pdays_bins = OrdinalEncoder(dtype=int)
df_bank_additional_full['pdays_recency_bins'] = encoder_pdays_bins.fit_transform(df_bank_additional_full[['pdays_recency_bins']])
print("'pdays_recency_bins' created and ordinal encoded.")
```

```python
# 3. Number of Unknowns
# This counts the number of features explicitly encoded as '2' (representing 'unknown') for each client.
# Note: This is limited to 'default', 'housing', 'loan' where 'unknown' was explicitly mapped to 2.
# For other ordinally encoded columns (job, marital, education etc.), the integer corresponding to 'unknown' is not directly identifiable from the current state without re-running the encoder.
unknown_coded_cols = ['default', 'housing', 'loan']

df_bank_additional_full['num_unknowns'] = df_bank_additional_full[unknown_coded_cols].apply(lambda row: (row == 2).sum(), axis=1)
print("'num_unknowns' feature created.")
```

```python
# Re-executing the fixed 'Age Group Categories' cell
# 4. Age Group Categories
# Segment 'age' into meaningful groups and then ordinal encode them.
# The 'age' column in df_bank_additional_full is already scaled. To create meaningful age groups,
# we should use the original unscaled 'age' values. We can get them from the pre-encoding DataFrame.

# Reload the original DataFrame to access unscaled 'age' values, assuming it's available.
# It's important to ensure df_bank_additional_full_pre_encoding has the same index as df_bank_additional_full
# after duplicate removal to avoid misalignment.

# Re-load df_bank_additional_full_pre_encoding from the original file path and apply duplicate removal
# to ensure it aligns with the current df_bank_additional_full for merging.
original_df_for_age = pd.read_csv(bank_additional_full_filepath, sep=';')
original_df_for_age.drop_duplicates(inplace=True)

# Get original age values corresponding to the current df_bank_additional_full's index
unscaled_age = original_df_for_age.loc[df_bank_additional_full.index, 'age']

min_original_age = unscaled_age.min() # This should be 17
max_original_age = unscaled_age.max() # This should be 98

age_bins = [min_original_age - 1, 25, 40, 55, max_original_age + 1] # Ensure bins cover all ages and are inclusive
age_labels = ['Youth (<=25)', 'Young Adult (26-40)', 'Middle-Aged (41-55)', 'Senior (>55)']

# Create the binned feature using unscaled_age
df_bank_additional_full['age_group'] = pd.cut(unscaled_age, bins=age_bins, labels=age_labels, right=True)

# Ordinal encode the new categorical 'age_group' column
encoder_age_group = OrdinalEncoder(dtype=int)
df_bank_additional_full['age_group'] = encoder_age_group.fit_transform(df_bank_additional_full[['age_group']])
print("'age_group' created and ordinal encoded.")
```

### **2.4 Handling Class Imbalance**


Target variable `y` ('yes'/'no' for term deposit subscription) is highly imbalanced, section will demonstrate 2 common techniques to address it namely **SMOTE (Synthetic Minority Over-sampling Technique)** & **Random Under-sampling**.

### 2.4.1 Original Class Distribution

Visualize original distribution of target variable `y` to clearly see imbalance.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualize original class distribution
plt.figure(figsize=(6, 5))
sns.countplot(x='y', data=df_bank_additional_full, palette='viridis', hue='y', legend=False)
plt.title('Original Class Distribution of Target Variable (y)')
plt.xlabel('Subscription (0=No, 1=Yes)')
plt.ylabel('Count')
plt.show()

print("Original Class Distribution:")
print(df_bank_additional_full['y'].value_counts())
print(f"Proportion of 'yes' (1): {df_bank_additional_full['y'].value_counts(normalize=True)[1]:.2%}")
```

### 2.4.2 SMOTE (Synthetic Minority Over-sampling Technique)
SMOTE is an over-sampling technique which generates synthetic samples from minority class, it works by selecting sample from minority class and then finding its k-nearest neighbors. Synthetic samples are generated along line segments joining minority sample and its selected neighbors, which helps to increase number of minority class samples, making dataset more balanced without simply duplicating existing samples.

```python
# Define features (X) and target (y)
X = df_bank_additional_full.drop('y', axis=1)
y = df_bank_additional_full['y']

# Split data into training and test sets before applying SMOTE to avoid data leakage
# For demonstration, we'll apply SMOTE on the full dataset, but in a real scenario,
# it should be applied only to the training set within cross-validation.

# Apply SMOTE
smote = SMOTE(random_state=42)
X_resampled_smote, y_resampled_smote = smote.fit_resample(X, y)

print("Class distribution after SMOTE:")
print(y_resampled_smote.value_counts())

# Visualize class distribution after SMOTE
plt.figure(figsize=(6, 5))
sns.countplot(x=y_resampled_smote, palette='viridis', hue=y_resampled_smote, legend=False)
plt.title('Class Distribution After SMOTE')
plt.xlabel('Subscription (0=No, 1=Yes)')
plt.ylabel('Count')
plt.show()
```

### 2.4.3 Random Under-sampling

Random Under-sampling is an under-sampling technique that aims to balance class distribution by randomly removing samples from majority class, which can be effective in reducing the number of samples in majority class to match minority class, but it carries risk of discarding potentially useful information from majority class.

```python
from imblearn.under_sampling import RandomUnderSampler

# Define features (X) and target (y) again (using original X and y before any resampling)
X = df_bank_additional_full.drop('y', axis=1)
y = df_bank_additional_full['y']

# Apply Random Under-sampling
rus = RandomUnderSampler(random_state=42)
X_resampled_rus, y_resampled_rus = rus.fit_resample(X, y)

print("Class distribution after Random Under-sampling:")
print(y_resampled_rus.value_counts())

# Visualize class distribution after Random Under-sampling
plt.figure(figsize=(6, 5))
sns.countplot(x=y_resampled_rus, palette='magma', hue=y_resampled_rus, legend=False)
plt.title('Class Distribution After Random Under-sampling')
plt.xlabel('Subscription (0=No, 1=Yes)')
plt.ylabel('Count')
plt.show()
```

#### Summary

This section demonstrated how **SMOTE** and **Random Under-sampling** can be used to rebalance the target variable distribution. Each method has its advantages and disadvantages:

*   **SMOTE** creates synthetic samples for the minority class, which can help retain information from the majority class but may introduce noise if not carefully applied.
*   **Random Under-sampling** reduces the majority class, leading to a balanced dataset but potentially losing valuable information. It's often quicker to train models on under-sampled data.

The choice between these or other techniques (like class weighting, which adjusts the cost of misclassification) depends on the specific dataset, the chosen model, and the business objective. These techniques will be integrated into the model training and tuning phases to ensure robust performance on the minority class.

### 2.2.5 Handling `duration` Feature (Data Leakage Mitigation)

The `duration` feature, representing last contact duration, is highly problematic due to its direct and unrealistic correlation with target variable (`y`). The outcome of call (subscription to term deposit) is only known *after* the call is made, and `duration = 0` implies `y = 'no'`. Including the feature directly in predictive model would lead to **data leakage**, resulting in an overly optimistic & untrustworthy model that cannot be used to predict subscription *before* a call is made.

To mitigate this data leakage, `duration` feature will be **removed** entirely from dataset, to ensure model's predictions are based on information available *before* contact, making it realistic and actionable tool for optimizing marketing campaigns.

```python
# Remove the 'duration' column to prevent data leakage
print(f"Shape of df_bank_additional_full before dropping 'duration': {df_bank_additional_full.shape}")
df_bank_additional_full.drop(columns=['duration'], inplace=True)
print(f"Shape of df_bank_additional_full after dropping 'duration': {df_bank_additional_full.shape}")

print("The 'duration' feature has been successfully removed.")
```

### **3.1 Evaluation Metrics Function**

To ensure consistent evaluation across all models, I'll define a helper function that calculates and displays the key performance metrics: Accuracy, Precision, Recall, F1-Score, and ROC-AUC, along with a confusion matrix. This function will be reused for each model.

```python
def evaluate_model(y_true, y_pred, y_prob, model_name):
    """Calculates and prints classification metrics and displays a confusion matrix."""

    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)
    roc_auc = roc_auc_score(y_true, y_prob)

    print(f"--- {model_name} Performance ---")
    print(f"Accuracy: {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall: {recall:.4f}")
    print(f"F1-Score: {f1:.4f}")
    print(f"ROC-AUC: {roc_auc:.4f}")

    print(f"\nClassification Report for {model_name}:")
    print(classification_report(y_true, y_pred))

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title(f'Confusion Matrix for {model_name}')
    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.show()

    # Calculate ROC curve data
    fpr, tpr, _ = roc_curve(y_true, y_prob)

    return accuracy, precision, recall, f1, roc_auc, fpr, tpr
```

It trains very quickly and directly outputs probabilities, which are useful for ranking potential customers. It provides a solid foundation against which more complex models can be compared, helping to determine if the added complexity of advanced models yields a significant performance improvement.

```python
# Initialize and train the Logistic Regression model
log_reg = LogisticRegression(random_state=42, solver='liblinear', class_weight='balanced') # 'class_weight' for imbalance
log_reg.fit(X_train, y_train)

# Make predictions on the validation set
y_pred_lr = log_reg.predict(X_val)
y_prob_lr = log_reg.predict_proba(X_val)[:, 1] # Probability of the positive class

# Evaluate the Logistic Regression model
print("\n--- Evaluating Logistic Regression (Baseline Model) ---")
evaluate_model(y_val, y_pred_lr, y_prob_lr, "Logistic Regression")
```

### **3.3 Comparison Model 1: Random Forest**

**Purpose and Importance:**

Random Forest is an ensemble learning method that constructs a multitude of decision trees during training and outputs the mode of the classes (for classification) or mean prediction (for regression) of the individual trees. It's well-suited for this dataset because of its robustness to overfitting, ability to handle mixed data types (numerical and categorical after encoding), and its capacity to provide valuable insights into feature importance. Its ensemble nature helps in capturing complex non-linear relationships without being overly sensitive to outliers.

```python
# Initialize and train the Random Forest model
# Using 'class_weight' to handle imbalance, similar to Logistic Regression baseline
rf_clf = RandomForestClassifier(random_state=42, class_weight='balanced')
rf_clf.fit(X_train, y_train)

# Make predictions on the validation set
y_pred_rf = rf_clf.predict(X_val)
y_prob_rf = rf_clf.predict_proba(X_val)[:, 1]

# Evaluate the Random Forest model
print("\n--- Evaluating Random Forest ---")
evaluate_model(y_val, y_pred_rf, y_prob_rf, "Random Forest")
```

### **3.4 Comparison Model 2: Gradient Boosting Machines (XGBoost)**

**Purpose and Importance:**

Gradient Boosting Machines, particularly implementations like XGBoost, are state-of-the-art ensemble methods that sequentially build decision trees, with each new tree correcting the errors of the previous ones. They are highly effective for tabular data, known for achieving high predictive accuracy, and are capable of capturing complex non-linear relationships and interactions between features. XGBoost also offers built-in mechanisms to handle class imbalance, making it a strong candidate for this problem. It often excels where other models might struggle with intricate patterns in the data.

```python
# Initialize and train the XGBoost model
# 'scale_pos_weight' is used to handle class imbalance. It is calculated as (count(negative_examples) / count(positive_examples))
# y_train.value_counts()[0] is count of negative class (0)
# y_train.value_counts()[1] is count of positive class (1)
scale_pos_weight_val = y_train.value_counts()[0] / y_train.value_counts()[1]

xgb_clf = XGBClassifier(random_state=42, use_label_encoder=False, eval_metric='logloss', scale_pos_weight=scale_pos_weight_val)
xgb_clf.fit(X_train, y_train)

# Make predictions on the validation set
y_pred_xgb = xgb_clf.predict(X_val)
y_prob_xgb = xgb_clf.predict_proba(X_val)[:, 1]

# Evaluate the XGBoost model
print("\n--- Evaluating XGBoost ---")
evaluate_model(y_val, y_pred_xgb, y_prob_xgb, "XGBoost")
```
